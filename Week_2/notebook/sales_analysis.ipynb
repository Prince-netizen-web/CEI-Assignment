```python
import pandas as pd
import sqlite3

df = pd.read_csv("../data/Sample - Superstore.csv", encoding='latin1')

df.head()
```

```python
conn = sqlite3.connect("superstore.db")

df.to_sql("sales",conn,if_exists="replace", index=False)
print("Database created successfully!")
```

```python
query = """
SELECT *
FROM sales
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT *
FROM sales
WHERE Region = 'West';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT [Row ID],[Order ID],[Customer ID],[Customer Name]
FROM sales
WHERE Category = 'Technology';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT *
FROM sales
WHERE Sales > 500;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT *
FROM sales
WHERE "Order Date" >= '2017-01-01';
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT Category,
       ROUND(SUM(Sales),2) AS Total_Sales
FROM sales
GROUP BY Category;
"""

pd.read_sql(query, conn)
```

```python
query = """
Select Region,
        Round(AVG(Profit),2) as Avg_Profit
        from sales
Group by Region;
"""
pd.read_sql(query,conn)
```

```python
query="""
select [Sub-Category], 
        Sum(Quantity) as total_quantity
from sales Group by "Sub-Category";
"""
pd.read_sql(query,conn)
```

```python
query = """
SELECT "Product Name",
       ROUND(SUM(Sales),2) AS Total_Sales
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query ="""
Select "Customer Name",
Sum(Profit) as total_profit
from sales 
group by "Customer Name"
order by total_profit desc
limit 5;
"""
pd.read_sql(query,conn)
```

```python
query = """
SELECT "Category",
       SUM(Profit) AS total_category_profit
FROM sales
GROUP BY "Category"
HAVING SUM(Profit) > 10000
ORDER BY total_category_profit DESC
LIMIT 5;
"""
pd.read_sql(query, conn)
```

```python
query ="""
SELECT substr("Order Date",1,7) as Month,
SUM(Sales) as Monthly_Sales
from sales
Group By Month
Order By Month asc;
"""
pd.read_sql(query,conn)
```
