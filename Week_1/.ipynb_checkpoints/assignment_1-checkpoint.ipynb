```python
df["seller_name"] = df["seller_name"].fillna(df["seller_name"].mode()[0])
```

```python
df["what_customers_said"] = df["what_customers_said"].fillna(df["what_customers_said"].mode()[0])
```

```python
df["seller_information"] = df["seller_information"].fillna(df["seller_information"].mode()[0])
```

```python
df["variations"] = df["variations"].fillna(df["variations"].mode()[0])
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df["discount"] = df["discount"].fillna(df["discount"].mean())
```
